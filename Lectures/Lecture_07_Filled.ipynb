## Creating NumPy Arrays
NumPy gives us access to a new data type: the array. Arrays are similar to lists, but differ in important ways:
* Arrays can be multidimensional. A 1D array is similar to a list, but we can also create 2D, 3D, nD arrays.
* All elements of an array must have the same data type.
* Arrays include lots of mathematical functionality (eg. operators, functions, methods) which isn't the case for lists.

You can convert a list to an array with the <code>array</code> function in NumPy. There are also other functions available to quickly create common arrays such as <code>arange</code>, <code>linspace</code>, <code>ones</code> and <code>zeros</code>.

````python
# creating a 1D array (ie. vector)
numbers_list = [1, 5, 2, 6, 8, 3]
numbers_array = np.array(numbers_list)
numbers_array
````

````python
# creating a 2D array (ie. matrix)
matrix_list = [[2, 1], [3, 5], [4, 6]]
matrix_array = np.array(matrix_list)
matrix_array
````

````python
# creating 10 x 10 zeros matrix
np.zeros([10, 10])
````

````python
# evenly spaced array with spacing of 2
np.arange(0, 10, 2)
````

````python
# evenly spaced array with 100 elements
np.linspace(0, 10, 100)
````

## Element-wise Processes
NumPy supports elementwise operations, functions and indexing, avoiding the need for loops! An element-wise process is one that "does something" for each element. This allows you to more easily and efficiently perform calculations with collections of values than you could with Python's in-built data structures (lists, tuples, etc.).

````python
# examples
````

````python
# eg. square each number in a list
# lists don't have elmentwise support, so need a for loop
squared_list = []
for number in numbers_list:
    squared_list.append(number**2)

squared_list
````

````python
# arrays do have elementwise support - don't need for loop
squared_array = numbers_array ** 2
squared_array
````

````python
# adding 5 to everything
numbers_array + 5
````

````python
# often work with arrays of the same length
squared_array - numbers_array
````

````python
# elementwise support is built into numpy functions
# getting cosine of integers from 0 to 9
integers = np.arange(0, 10)
np.cos(integers)
````

````python
# math module doesn't have elementwise support
import math
math.cos(integers)
````

````python
# relational operators are elementwise in numpy
np.array([4, 0, 1]) == np.array([4, 0, 2])
````

````python
# not elementwise with lists
[4, 0, 1] == [4, 0, 2]
````

## Indexing and Slicing
NumPy arrays can be indexed and sliced like lists. 

In addition to slicing, it will allow for a sequence of indexes to be provided to select a subset of an array, meaning you can select any subset and not just adjacent elements.

Further functionality is also available for higher dimensional arrays, since arrays understand the concept of dimensionality. For example, in a matrix you could index with:
```{python}
matrix_name[row_index, col_index]
```

````python
# arrays have list indexing functionality
numbers_array[0] # first element
````

````python
numbers_array[-1] # last element
````

````python
numbers_array[2:5] # elements 3-5
````

````python
# have new functionality for higher-dimensional array
# can provide an index or slice for each dimension
matrix_array[1, 0]
````

````python
matrix_array[:2, 1] # first 2 rows in the 2nd column
````

````python
matrix_array[0, :] # select all columns in 1st row
````

````python
matrix_array[:, -1] # select the last column
````

````python
# can use lists/arrays of indexes
numbers_array[[2, 0, 5]] # select 3rd, 1st and 5th element
````

````python
# can select conditionally with boolean indexing
numbers_array[numbers_array > 2] # select elements where condition (numbers > 2) is True
````

## Mathematics with NumPy
NumPy has functionality to support more advanced mathematics beyond the math module. This includes statistics, linear algebra and numerical methods.

### Example: Numerical methods
An underdamped motor lifting a mass produces a velocity profile given by the equation $v(t) = 1-e^{-t}\cos(10t)$ for $0 < t < 10$. 
1. Visualise the velocity with a plot.
2. Numerically compute the acceleration, noting $a = \cfrac{dv}{dt}$.
3. Numerically compute the position, noting $s = \int v \ dt$.
4. Visualise the position, velocity and acceleration behaviour in a single plot.

````python
# step 1
times = np.linspace(0, 10, 1000)
velocities = 1 - np.exp(-times) * np.cos(10 * times)

fig, ax = plt.subplots() # create blank figure and axes
ax.plot(times, velocities) # plot onto axes
ax.set_xlabel('time')
ax.set_ylabel('velocity')
plt.show()
````

````python
# step 2
dv = np.diff(velocities)
dt = np.diff(times)
accelerations = dv / dt
````

````python
# step 3
position_changes = velocities[:-1] * dt
positions = np.cumsum(position_changes)
````

````python
# step 4
fig, ax = plt.subplots(3, 1) # create blank figure and axes
ax[0].plot(times[:-1], positions)
ax[0].set_ylabel('position')

ax[1].plot(times, velocities) # plot onto axes
ax[1].set_ylabel('velocity')

ax[2].plot(times[:-1], accelerations)
ax[2].set_xlabel('time')
ax[2].set_ylabel('acceleration')
plt.show()
````
