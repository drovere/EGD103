## B5: Differentiating vectors (2 marks)
Engineers are often analysing the kinematics of mechanisms such as velocities and accelerations. In order to do this, you will need to be able to differentiate positions of the mechanism with respect to time. This can be done through numeric differentiation.

To differentiate a vector (x, y) with respect to time, you can do the following:
1. Find the components of the vector $x_n$ and $y_n$.
2. Find the components of the vector after one time step $x_{n+1}$ and $y_{n+1}$.
3. Computer the derivative of ($x_n$, $y_n$) as $\left(\cfrac{x_{n+1} - x_n}{\Delta t}, \cfrac{y_{n+1} - y_n}{\Delta t}\right)$
